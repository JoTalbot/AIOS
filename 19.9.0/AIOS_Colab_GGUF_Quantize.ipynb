# 🗜️ AIOS GGUF / AWQ Quantization

Конвертация слитой LoRA-модели (или базовой) в компактный **GGUF** для быстрого инференса на VPS через llama.cpp / Ollama (порт 11434 уже слушает).

```python
!pip install -q transformers torch
import torch, os
print('✅ Зависимости установлены')
```

```python
# === ЯЧЕЙКА 2: Загрузка слитой модели ===
from transformers import AutoModelForCausalLM, AutoTokenizer
model = AutoModelForCausalLM.from_pretrained('merged_model', torch_dtype=torch.float16, device_map='auto')
tokenizer = AutoTokenizer.from_pretrained('merged_model')
print('✅ Модель загружена')
```

```python
# === ЯЧЕЙКА 3: Экспорт в safetensors для llama.cpp ===
!pip install -q huggingface_hub
os.makedirs('export_gguf', exist_ok=True)
model.save_pretrained('export_gguf')
tokenizer.save_pretrained('export_gguf')
print('✅ Экспортировано в export_gguf/')
print('Дальше в Colab: pip install llama-cpp-python и конвертация через convert_hf_to_gguf.py')
```

```python
# === ЯЧЕЙКА 4: Конвертация в GGUF (llama.cpp) ===
!git clone -q --depth 1 https://github.com/ggerganov/llama.cpp
!python llama.cpp/convert_hf_to_gguf.py export_gguf --outfile aios_model.gguf --outtype q8_0
print('✅ Готово: aios_model.gguf')
print('Загрузите aios_model.gguf на VPS и запустите Ollama / llama.cpp.')
```
